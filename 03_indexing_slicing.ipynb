## 1️⃣ Indexing

Indexing is used to access a specific element of a NumPy array.

NumPy uses **zero-based indexing** — the first element has index `0`.

```python
scores = np.array([83, 95, 40, 99, 18, 72])
scores[0]
```

```text
83
```


````python
scores = np.array([83, 95, 40, 99, 18, 72])

print(scores[0])
print(scores[3])
````

````python
print(scores[1])
print(scores[4])
print(scores[5])
````

> [!TIP]
> NumPy indexing starts from `0` — the first element is `array[0]`, not `array[1]`.


## 2️⃣ Negative Indexing

Negative indexing accesses elements from the end of the array. The last element is `-1`, the second-to-last is `-2`, and so on.

```python
scores = np.array([83, 95, 40, 99, 18, 72])
scores[-1]
```

```text
72
```


````python
print(scores[-1])
print(scores[-2])
````

````python
print(scores[-6])
````

> [!TIP]
> Positive indexing starts at `0` from the front; negative indexing starts at `-1` from the back. Since `scores` has 6 elements, `-6` wraps all the way around to the *same* element as `0` — the first one.


## 3️⃣ Slicing

Slicing extracts a portion of a NumPy array.

**Syntax**

```python
array[start:stop:step]
```

- `start` — starting index, included.
- `stop` — ending index, **not included**.
- `step` — the interval between elements.

```python
scores = np.array([83, 95, 40, 99, 18, 72])
scores[1:4]
```

```text
[95 40 99]
```

Indexes `1`, `2`, and `3` are included; index `4` is not.


````python
print(scores[0:3])
````

### Omitting `stop`

If `stop` is omitted, the slice continues to the end of the array.

```python
scores[2:]
```
```text
[40 99 18 72]
```


````python
print(scores[1:])
````

````python
print(scores[:])
````

> [!TIP]
> `array[start:]` → starts at `start`, continues to the end. `array[:]` → a full copy of the array, start to end.


## 4️⃣ Slicing with `step`

`step` sets how many positions to move between selected elements.

```python
array[start:stop:step]
```

```python
scores[0:6:2]
```
```text
[83 40 18]
```

Every second element is selected.


````python
print(scores[0:8:2])
````

> [!NOTE]
> `scores` only has 6 elements, but `stop=8` above doesn't raise an error — NumPy automatically clips a slice's `stop` to the array's length. The result is identical to `scores[0:6:2]`.


````python
print(scores[::2])  # without start, stop
````

````python
print(scores[::-1])
````

> [!TIP]
> `step=2` → take every second element. `step=-1` with no `start`/`stop` reverses the whole array.


## 5️⃣ Modifying Array Elements

NumPy lets you change an element by assigning a new value to its index.

```python
scores[2] = 100
print(scores)
```
```text
[83 95 100 99 18 72]
```


````python
scores = np.array([83, 95, 40, 99, 18, 72])

scores[2] = 100
print(scores)
````

````python
scores[4] = 80
print(scores)
````

> [!TIP]
> `array[index] = new_value` → changes the value at that index in place. `scores[5]` still holds `72` here — only index `4` was reassigned to `80`.


## 6️⃣ 2D Array Indexing

A 2D array has rows and columns. Access a single element with two indexes:

```python
array[row, column]
```

Both start from `0`.

```python
matrix[0, 1]
```
```text
20
```


````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix)
````

````python
print(matrix[2, 1])
````

> [!TIP]
> `array[row, column]` → accesses one element in a 2D array. The first index picks the row, the second picks the column.


## 7️⃣ Selecting Rows and Columns from a 2D Array

An entire **row** is selected with a single index:

```python
matrix[1]
```
```text
[40 50 60]
```

An entire **column** is selected with `:` for rows and the column index:

```python
matrix[:, 1]
```
```text
[20 50 80]
```


````python
print(matrix[2])
````

````python
print(matrix[:, 1])
````

````python
print(matrix[:, 2])
````

> [!TIP]
> `array[row]` → an entire row. `array[:, column]` → an entire column, where `:` means "all rows."


## 8️⃣ 2D Array Slicing

Slicing selects multiple rows and columns at once:

```python
array[row_start:row_stop, column_start:column_stop]
```

`stop` is not included, just like 1D slicing.

```python
matrix[0:2, 1:3]
```
```text
[[20 30]
 [50 60]]
```

This selects rows `0`–`1` and columns `1`–`2`.


````python
print(matrix[0:2, 1:3])
````

````python
print(matrix[1:3, 1:3])
````

> [!TIP]
> For a 2D array, the first slice picks rows and the second picks columns — `array[rows, columns]`.


## 9️⃣ Modifying Multiple Elements

Slicing can modify several elements at once.

```python
matrix[0, :] = 100
```

This sets every element in the first row to `100`:

```text
[[100 100 100]
 [ 40  50  60]
 [ 70  80  90]]
```

`:` selects every column in that row.


````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

matrix[0, :] = 100
print(matrix)
````

````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

matrix[2, :] = 0
print(matrix)
````

> [!TIP]
> `array[row, :] = value` → overwrites every element in that row. The same works for a column with `array[:, col] = value`.


## 🧪 Extra Practice — Combining Indexing and Slicing

A few more reads on a fresh copy of `matrix`, mixing single-element indexing, row/column slicing, and reversed selections.


````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix[1, 1])
````

````python
print(matrix[0])
````

````python
print(matrix[0:3, 2:])
````

> [!NOTE]
> `matrix[:, 2]` and `matrix[:, 2:]` look similar but return different shapes: a single-index column collapses to 1D, while a *sliced* column keeps its 2D shape.


````python
print(matrix[:, 2])   # shape (3,)  — 1D
````

````python
matrix[:, 2:]         # shape (3, 1) — 2D
````

````python
print(matrix[1:, 1:])
````

> [!TIP]
> `array[:, col]` → a 1D array of that column's values. `array[:, col:]` → the same values, but kept as a 2D column.
