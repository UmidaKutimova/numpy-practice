## 1️⃣ Creating Arrays

`np.array()` builds a NumPy array from a Python list — the basic building block for everything that follows.


````python
array1 = np.array([22, 4, 63, 98, 2, 5, 33, 21])

print(array1)
````

> [!TIP]
> `np.array(list)` → converts a Python list into a NumPy array, which supports fast, vectorized operations that plain lists don't.


## 2️⃣ Number of Dimensions — `.ndim`

`.ndim` shows how many dimensions (axes) a NumPy array has. A plain list of numbers is one-dimensional, so `.ndim` is `1`.


````python
scores = np.array([83, 95, 40, 99, 18, 72])
print(scores)
````

````python
print(scores.ndim)
````

> [!TIP]
> `array.ndim` → number of dimensions. A 1D array (a simple list of values) has `ndim = 1`; a 2D array (rows and columns) has `ndim = 2`.


## 3️⃣ Shape — `.shape`

`.shape` returns the size of the array along each dimension, as a tuple.

- A 1D array with 6 elements → `(6,)`
- A 2D array with 2 rows and 3 columns → `(2, 3)`


````python
print(scores.shape)
````

> [!TIP]
> `array.shape` → a tuple showing the length of each dimension.


## 4️⃣ Total Element Count — `.size`

`.size` returns the total number of elements in the array, regardless of how many dimensions it has.


````python
print(scores.size)
````

```python
scores = np.array([83, 95, 40, 99, 18, 72])
print(scores.size)
```
```text
6
```

The array holds **6 elements**, so `size` is `6`.


> [!TIP]
> `array.size` → total number of elements, no matter the shape. A `(2, 3)` array and a `(6,)` array can both have `size = 6`.


## 5️⃣ Data Type — `.dtype`

`.dtype` shows the data type of the elements stored in the array.

Common NumPy data types:

| Type | Meaning |
|------|---------|
| `int` | integer numbers |
| `float` | decimal numbers |
| `bool` | `True` / `False` |
| `str` | text |


````python
print(scores.dtype)
````

````python
prices = np.array([12.5, 20.99, 35.75])
print(prices.dtype)
````

> [!TIP]
> `array.dtype` → the data type of the elements. Mixing even one decimal number into an otherwise integer array upgrades the whole array to `float64`.


## 6️⃣ 2D Arrays

A 2D array is built from a list of lists — each inner list becomes a row.


````python
scores_2d = np.array([
    [83, 95, 40],
    [99, 18, 72]
])

print(scores_2d.shape)
````

````python
print(scores_2d.size)
````

````python
print(scores_2d.dtype)
````

> [!NOTE]
> Even though `scores_2d` has 2 rows and 3 columns, its `.size` is still `6` — `.size` always counts every element, not just the rows.


## 🧪 Mini Practice — Inspecting an Array

Bringing `.ndim`, `.shape`, `.size`, and `.dtype` together on one dataset:


````python
temperatures = np.array([33, 21, 25, 40, 38, 37, 36])

print(temperatures.ndim)
print(temperatures.shape)
print(temperatures.size)
print(temperatures.dtype)
````

Adding just two decimal values to the same kind of data changes the dtype from `int64` to `float64`:


````python
temperatures = np.array([33.3, 21, 25, 40, 38, 37, 36.5])
print(temperatures.dtype)
````
