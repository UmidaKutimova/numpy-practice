## 1️⃣ `np.concatenate()`

`np.concatenate()` joins two or more NumPy arrays into one.

For 1D arrays, it joins them along the only axis they have.


```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))

print(result)
```

```python
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

result = np.concatenate((a, b))

print(result)
```

### Concatenating 2D arrays along an axis

For 2D arrays, `axis` controls the direction of the join:

- `axis=0` (the default) — stacks along rows, adding more rows.
- `axis=1` — stacks along columns, adding more columns.


```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

result = np.concatenate((a, b))

print(result)
```

```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

result = np.concatenate((a, b), axis=1)

print(result)
```

> [!TIP]
> `np.concatenate((a, b))` defaults to `axis=0`. Pass `axis=1` to join side by side instead of stacking on top of each other — both arrays must have compatible shapes along the axis you're *not* joining on.


## 2️⃣ `np.vstack()`

`np.vstack()` joins arrays **vertically**, stacking them on top of each other and increasing the number of rows.

For 2D arrays, it works the same as `np.concatenate(..., axis=0)`.


```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

result = np.vstack((a, b))

print(result)
```

```python
a = np.array([
    [10, 20],
    [30, 40]
])

b = np.array([
    [50, 60],
    [70, 80]
])

result = np.vstack((a, b))

print(result)
```

> [!TIP]
> `np.vstack((a, b))` → same result as `np.concatenate((a, b), axis=0)`, but arguably reads more clearly when the intent is specifically "stack these vertically."


## 3️⃣ `np.hstack()`

`np.hstack()` joins arrays **horizontally**, placing them side by side and increasing the number of columns.

For 2D arrays, it works the same as `np.concatenate(..., axis=1)`.


```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

result = np.hstack((a, b))

print(result)
```

> [!TIP]
> `np.hstack((a, b))` → same result as `np.concatenate((a, b), axis=1)`. Together, `vstack`/`hstack` are shorthand for the two most common concatenation directions.
