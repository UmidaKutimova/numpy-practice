<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">➕ np.sum()</h2>
</div>


`np.sum()` calculates the sum of all elements in a NumPy array.

**Example**

```python
numbers = np.array([10, 20, 30, 40])
np.sum(numbers)
```

**Output**

```text
100
```

Calculation: `10 + 20 + 30 + 40 = 100`


````python
sales = np.array([120, 250, 180, 300, 150])

print(sales.sum())
````

````python
sales = np.array([120, 250, 180, 300, 150])

print(np.sum(sales))
````

<div style="background:#eafaf1; border-left:4px solid #27ae60; padding:10px 18px; border-radius:0 6px 6px 0; margin-top:10px;">
<b>💡 Key point:</b> <code>np.sum(array)</code> → returns the total sum of the elements.
</div>


<a id="mean"></a>
<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">📐 np.mean()</h2>
</div>


`np.mean()` calculates the arithmetic mean, or average, of the elements in a NumPy array — the sum of all values divided by the number of values.

**Example**

```python
numbers = np.array([10, 20, 30, 40])
np.mean(numbers)
```

**Output**

```text
25.0
```

Calculation: `(10 + 20 + 30 + 40) / 4 = 25`


````python
sales = np.array([120, 250, 180, 300, 150])

print(sales.mean())
````

<div style="background:#eafaf1; border-left:4px solid #27ae60; padding:10px 18px; border-radius:0 6px 6px 0; margin-top:10px;">
<b>💡 Key point:</b> <code>np.mean(array)</code> → returns the average value of the array.
</div>


<a id="min"></a>
<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">🔽 np.min()</h2>
</div>


`np.min()` returns the smallest value in a NumPy array.

**Example**

```python
numbers = np.array([10, 25, 5, 40, 15])
np.min(numbers)
```

**Output**

```text

```


````python
sales = np.array([120, 250, 180, 300, 150])

print(sales.min())
````

<div style="background:#eafaf1; border-left:4px solid #27ae60; padding:10px 18px; border-radius:0 6px 6px 0; margin-top:10px;">
<b>💡 Key point:</b> <code>np.min(array)</code> → returns the minimum value.

<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">🔼 np.max()</h2>

`np.max()` returns the largest value in a NumPy array.

**Example**

```python
numbers = np.array([10, 25, 5, 40, 15])
np.max(numbers)
```

<a id="median"></a>
<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">🎯 np.median()</h2>
</div>


`np.median()` returns the middle value of a sorted array. If the array has an odd number of values, the median is the exact middle value.

**Example**

```python
numbers = np.array([10, 30, 20, 50, 40])
np.median(numbers)
```

Sorted values: `[10, 20, 30, 40, 50]` → middle value is `30`

**Output**

```text
30.0
```


````python
sales = np.array([120, 250, 180, 300, 150])

print(np.median(sales))
````

<div style="background:#eafaf1; border-left:4px solid #27ae60; padding:10px 18px; border-radius:0 6px 6px 0; margin-top:10px;">
<b>💡 Key point:</b> <code>np.median(array)</code> → returns the middle value after sorting the data.
</div>


<a id="std"></a>
<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">📏 np.std()</h2>
</div>


`np.std()` calculates the standard deviation of an array — how much the values differ from the mean.

- A **small** standard deviation → values are close to the mean.
- A **large** standard deviation → values are more spread out.


````python
sales = np.array([120, 250, 180, 300, 150])

print(np.std(sales))
````

<div style="background:#eafaf1; border-left:4px solid #27ae60; padding:10px 18px; border-radius:0 6px 6px 0; margin-top:10px;">
<b>💡 Key point:</b> <code>np.std(array)</code> → returns the standard deviation.
</div>


<a id="var"></a>
<div style="border-left:6px solid #3498db; background:#f4f8fb; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">📊 np.var()</h2>
</div>


`np.var()` calculates the variance of an array — how spread out the values are from the mean.

Variance is the square of the standard deviation.


````python
sales = np.array([120, 250, 180, 300, 150])

print(np.var(sales))
````

<div style="background:#eafaf1; border-left:4px solid #27ae60; padding:10px 18px; border-radius:0 6px 6px 0; margin-top:10px;">
<b>💡 Key point:</b> <code>np.var(array)</code> → returns the variance.
</div>


<a id="practice"></a>
<div style="border-left:6px solid #e67e22; background:#fff6ec; padding:14px 22px; border-radius:0 8px 8px 0; margin-top:10px;">
<h2 style="color:#182848; margin:0;">🧪 Mini Practice — Sales Statistics</h2>
</div>


Bringing all seven functions together on one dataset:


````python
sales = np.array([120, 250, 180, 300, 150, 220, 280, 190])

print("Total: ", sales.sum())
print("Mean: ", sales.mean())
print("Min: ", sales.min())
print("Max: ", sales.max())
print("Median: ", np.median(sales))
print("Standard deviation: ", np.std(sales))
print("Variance: ", np.var(sales))
````
