````python
import numpy as np
````

## 1️⃣ `np.zeros()`

`np.zeros()` creates a NumPy array filled with zeros.

```python
zeros_array = np.zeros(5)
```

```text
[0. 0. 0. 0. 0.]
```

The argument `5` specifies the number of elements.


````python
zeros_array = np.zeros(5)
print(zeros_array)
````

````python
print(zeros_array.shape)
print(zeros_array.size)
print(zeros_array.dtype)
````

> [!TIP]
> `np.zeros(n)` → creates an array of `n` zeros. Note the `dtype` defaults to `float64`, even though the values look like whole numbers.


## 2️⃣ `np.ones()`

`np.ones()` creates a NumPy array filled with ones.

```python
ones = np.ones(5)
```

```text
[1. 1. 1. 1. 1.]
```

The argument `5` specifies the number of elements.


````python
ones = np.ones(5)
print(ones)
````

> [!TIP]
> `np.ones(n)` → creates an array of `n` ones.


## 3️⃣ `np.arange()`

`np.arange()` creates a NumPy array containing evenly spaced values within a specified interval.

**Syntax**

```python
np.arange(start, stop, step)
```

- `start` — the starting value.
- `stop` — the ending boundary. This value is **not included**.
- `step` — the difference between consecutive values.


````python
numbers = np.arange(5)
print(numbers)
````

````python
numbers2 = np.arange(2, 10)
print(numbers2)
````

````python
numbers3 = np.arange(2, 10, 2)
print(numbers3)
````

````python
num = np.arange(10, 50, 5)
print(num)
````

> [!TIP]
> `np.arange(start, stop, step)` → generates values from `start` up to (but not including) `stop`, spaced by `step`. With one argument, `start` defaults to `0` and `step` to `1`.


## 4️⃣ `np.random.randint()`

`np.random.randint()` creates an array containing random integers from a specified range.

**Syntax**

```python
np.random.randint(low, high, size)
```

- `low` — the lowest possible value, included.
- `high` — the upper boundary, **not included**.
- `size` — the number or shape of values to generate.

The output changes every run since the values are randomly generated.


````python
np.random.randint(1, 10, size=5)
````

````python
# 2D array
np.random.randint(1, 10, size=(3, 4))
````

> [!TIP]
> `np.random.randint(low, high, size)` → random integers from `low` (included) up to `high` (excluded). Pass a tuple like `size=(3, 4)` to generate a 2D array.


## 5️⃣ `np.random.random()`

`np.random.random()` generates random floating-point numbers between `0` (inclusive) and `1` (exclusive).

**Syntax**

```python
np.random.random(size)
```

- `size` — the number or shape of values to generate.
- Every value is `>= 0` and `< 1`.


````python
random_numbers = np.random.random(5)
print(random_numbers)
````

````python
random_numbers = np.random.random(10)
print(random_numbers)
````

> [!TIP]
> `np.random.random(size)` → random decimals in `[0, 1)`. Pass a tuple, e.g. `size=(2, 3)`, for a 2D array.
